```python
from pybis import Openbis
o = Openbis("https://playground.datastore.bam.de/")

o.set_token("your_token_here")
```

```python
space = o.get_space('X.1_MATERIALS')
space.get_projects()
```

```python
chemicals_project = space.new_project(
    code        = 'CHEMICALS_SYNTHESIS',#This is the name of the project
    description = 'Chemicals used in the "Material Synthesis" tutorial example. The records are fictional and intended solely for training and demonstration purposes.//Chemikalien aus dem Tutorial Beispiel „Material Synthesis“. Die Einträge sind fiktiv und dienen ausschließlich Schulungs- und Demonstrationszwecken.'
)
chemicals_project.save()
```

```python
samples_project = space.new_project(
    code        = 'SAMPLES_SYNTHESIS',#This is the name of the project
    description = 'Samples generated in the "Material Synthesis" tutorial example. The records are fictional and intended solely for training and demonstration purposes.//Proben aus dem Tutorial Beispiel "Material Synthesis". Die Einträge sind fiktiv und dienen ausschließlich Schulungs- und Demonstrationszwecken.'
)
samples_project.save()
```
